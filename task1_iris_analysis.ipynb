```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
print(df.shape)
print(df.columns)
print(df.head())
print(df.describe())
```

```python
plt.figure(figsize = (8, 5))
sns.scatterplot(data = df, x='sepal_length', y = 'petal_length', hue = 'species')
plt.title('Sepal Length vs Petal Length')
plt.show()
```

```python
df['sepal_length'].hist(bins=20, color='steelblue', edgecolor='black')
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.boxplot(data=df, x='species', y= 'petal_length')
plt.title('Petal Length by Species')
plt.show()
```

The Scatter Plot shows that setosa flowers have shorter petals. The histogram shows sepal length is roughly normally distributed. Box plots reveal versicolor and virginica have some overlap in petal length.
